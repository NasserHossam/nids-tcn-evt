```python
from collections import Counter

counts = Counter()
total = 0
for chunk in pd.read_csv(path, usecols=["Attack"], chunksize=500_000):
    counts.update(chunk["Attack"].value_counts().to_dict())
    total += len(chunk)

print("Total flows:", total)
for k, v in counts.most_common():
    print(f"{k:20s} {v:>10,}  ({v/total:.2%})")
```

```python
from collections import Counter

order = []
hosts_all, hosts_benign = Counter(), Counter()

for i, chunk in enumerate(pd.read_csv(path, usecols=["Attack", "IPV4_SRC_ADDR"], chunksize=500_000)):
    is_b = chunk["Attack"] == "Benign"
    order.append((i, round(is_b.mean(), 3), chunk["Attack"].nunique()))
    hosts_all.update(chunk["IPV4_SRC_ADDR"].value_counts().to_dict())
    hosts_benign.update(chunk.loc[is_b, "IPV4_SRC_ADDR"].value_counts().to_dict())

print("Per 500k-row block: (block, benign fraction, number of classes)")
for row in order:
    print(row)

print("\nUnique source IPs:", len(hosts_all))
print("Source IPs with benign flows:", len(hosts_benign))
print("Top 10 hosts by benign flows:", hosts_benign.most_common(10))
```

```python
import numpy as np

b = np.array(sorted(hosts_benign.values(), reverse=True))
total_b = b.sum()
print("Total benign flows:", total_b)   # should be 6,099,469

for k in [1, 5, 10, 50, 100]:
    print(f"top {k} hosts hold {b[:k].sum() / total_b:.1%} of benign flows")

for n in [10, 100, 1000, 10000]:
    print(f"hosts with >= {n} benign flows:", int((b >= n).sum()))
```
